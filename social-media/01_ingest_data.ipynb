```python
from pyspark.sql import Row

data = [
    Row(
        post_id="1",
        platform="twitter",
        user_id="u001",
        username="john",
        content="AI is changing the world",
        created_at="2026-10-27"
    )
]

df = spark.createDataFrame(data)

df.write.mode("append") \
.saveAsTable("social_media.bronze.posts")
```

```python
verify=spark.sql("select * from social_media.bronze.posts")
display(verify)
```
